```python
# Cell 2 — Build clean client datasets for federated training
# This does NOT use FL_MU.csv and does NOT merge raw client data for FL training.

from pathlib import Path
import json
import gc
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split

SEED = 42
rng = np.random.default_rng(SEED)

DATA_DIR = Path(
    "/root/.cache/kagglehub/datasets/linthoi1/fl-mu-dataset/versions/1/FL-MU-Dataset/FL-MU"
)
RUN_DIR = Path("/content/flmu_prepared")
RUN_DIR.mkdir(exist_ok=True)

CLIENT_FILES = {
    "C1": DATA_DIR / "C1_data.csv",
    "C2": DATA_DIR / "C2_data.csv",
    "C3": DATA_DIR / "C3_data.csv",
}

# Counts obtained from your completed audit.
TOTAL_ROWS = {
    "C1": 4_415_331,
    "C2": 17_251_169,
    "C3": 2_617_850,
}

# Semantic mapping reconstructed from the FL-MU client attack allocation.
CATEGORY_TO_SEMANTIC = {
    "C1": {0: "DDoS", 1: "Sinkhole", 2: "MiTM", 3: "Normal"},
    "C2": {0: "DDoS", 1: "DoS", 2: "Normal", 3: "SelectiveForwarding"},
    "C3": {0: "DDoS", 1: "DoS", 2: "Normal", 3: "Sybil"},
}

GLOBAL_CLASSES = [
    "Normal",
    "DDoS",
    "DoS",
    "MiTM",
    "SelectiveForwarding",
    "Sinkhole",
    "Sybil",
]
CLASS_TO_ID = {name: index for index, name in enumerate(GLOBAL_CLASSES)}

# Keep the natural class imbalance. Do NOT balance by sampling.
# Class-weighted focal loss will be used during local training.
MAX_ROWS_PER_CLIENT = 150_000
CHUNK_SIZE = 250_000
TEST_SIZE = 0.15
VALIDATION_SIZE = 0.15
Z_SCORE_CLIP = 10.0

# Target/metadata fields are forbidden as model inputs.
FORBIDDEN_COLUMNS = {"label", "category", "type", "FPT", "LPT"}

header = pd.read_csv(CLIENT_FILES["C1"], nrows=0)
FEATURE_COLUMNS = [
    column for column in header.columns
    if column not in FORBIDDEN_COLUMNS
]

assert len(FEATURE_COLUMNS) == 70, (
    f"Expected 70 usable features after dropping labels and FPT/LPT; "
    f"found {len(FEATURE_COLUMNS)}."
)

USE_COLUMNS = FEATURE_COLUMNS + ["category"]
print(f"Using {len(FEATURE_COLUMNS)} candidate features.")
print("Excluded:", sorted(FORBIDDEN_COLUMNS))


def stream_uniform_sample(client_name, csv_path):
    """Uniformly samples approximately MAX_ROWS_PER_CLIENT records without loading all data."""
    probability = min(1.0, MAX_ROWS_PER_CLIENT / TOTAL_ROWS[client_name])
    pieces = []

    for chunk in pd.read_csv(
        csv_path,
        usecols=USE_COLUMNS,
        chunksize=CHUNK_SIZE,
        low_memory=False
    ):
        keep = rng.random(len(chunk)) < probability
        if keep.any():
            pieces.append(chunk.loc[keep])

    sampled = pd.concat(pieces, ignore_index=True)

    # Make the final cap exact while preserving an approximately uniform sample.
    if len(sampled) > MAX_ROWS_PER_CLIENT:
        sampled = sampled.sample(
            n=MAX_ROWS_PER_CLIENT,
            random_state=SEED
        ).reset_index(drop=True)

    semantic = sampled["category"].map(CATEGORY_TO_SEMANTIC[client_name])

    if semantic.isna().any():
        unknown = sorted(sampled.loc[semantic.isna(), "category"].unique().tolist())
        raise ValueError(f"{client_name}: unmapped category values: {unknown}")

    y = semantic.map(CLASS_TO_ID).to_numpy(dtype=np.int64)

    # Convert all selected network-flow features to numbers.
    X = (
        sampled[FEATURE_COLUMNS]
        .apply(pd.to_numeric, errors="coerce")
        .to_numpy(dtype=np.float32)
    )

    print(f"\n{client_name}: retained {len(sampled):,} records")
    print(
        pd.Series(semantic)
        .value_counts()
        .rename_axis("semantic_class")
        .reset_index(name="records")
        .to_string(index=False)
    )

    del pieces, sampled, semantic
    gc.collect()
    return X, y


def local_split(X, y, client_name):
    """Creates train/validation/test splits without cross-client mixing."""
    class_counts = pd.Series(y).value_counts()

    if class_counts.min() < 7:
        raise ValueError(
            f"{client_name}: a class has fewer than 7 sampled records. "
            "Increase MAX_ROWS_PER_CLIENT before splitting."
        )

    indices = np.arange(len(y))

    train_idx, temporary_idx = train_test_split(
        indices,
        test_size=TEST_SIZE + VALIDATION_SIZE,
        random_state=SEED,
        stratify=y,
    )

    temporary_y = y[temporary_idx]
    validation_fraction_of_temporary = VALIDATION_SIZE / (TEST_SIZE + VALIDATION_SIZE)

    val_idx, test_idx = train_test_split(
        temporary_idx,
        test_size=1 - validation_fraction_of_temporary,
        random_state=SEED,
        stratify=temporary_y,
    )

    return {
        "train": (X[train_idx], y[train_idx]),
        "val": (X[val_idx], y[val_idx]),
        "test": (X[test_idx], y[test_idx]),
    }


# 1. Read each real FL client independently and sample it.
raw_splits = {}

for client_name, csv_path in CLIENT_FILES.items():
    X_client, y_client = stream_uniform_sample(client_name, csv_path)
    raw_splits[client_name] = local_split(X_client, y_client, client_name)
    del X_client, y_client
    gc.collect()


# 2. Federated-compatible global standardization:
#    only aggregate per-feature count, sum, and squared sum from CLIENT TRAINING SPLITS.
feature_count = len(FEATURE_COLUMNS)
sum_values = np.zeros(feature_count, dtype=np.float64)
sum_squares = np.zeros(feature_count, dtype=np.float64)
valid_counts = np.zeros(feature_count, dtype=np.float64)

for client_name in CLIENT_FILES:
    X_train, _ = raw_splits[client_name]["train"]
    finite = np.isfinite(X_train)

    valid_counts += finite.sum(axis=0)
    sum_values += np.where(finite, X_train, 0.0).sum(axis=0, dtype=np.float64)
    sum_squares += np.where(finite, X_train * X_train, 0.0).sum(axis=0, dtype=np.float64)

if np.any(valid_counts == 0):
    empty_features = [
        FEATURE_COLUMNS[index]
        for index in np.where(valid_counts == 0)[0]
    ]
    raise ValueError(f"Features with no finite training values: {empty_features}")

mean = sum_values / valid_counts
variance = np.maximum((sum_squares / valid_counts) - (mean ** 2), 0.0)
std = np.sqrt(variance)

# Remove constant/near-constant features based on training data only.
keep_mask = std > 1e-12
SELECTED_FEATURES = [
    feature for feature, keep in zip(FEATURE_COLUMNS, keep_mask) if keep
]

print(f"\nRemoved {len(FEATURE_COLUMNS) - len(SELECTED_FEATURES)} constant features.")
print(f"Prepared feature count: {len(SELECTED_FEATURES)}")

mean = mean[keep_mask]
std = std[keep_mask]


def normalize(X):
    X = X[:, keep_mask].astype(np.float32, copy=True)
    X = np.where(np.isfinite(X), X, mean[None, :])
    X = (X - mean[None, :]) / std[None, :]
    return np.clip(X, -Z_SCORE_CLIP, Z_SCORE_CLIP).astype(np.float32)


# 3. Normalize and persist each local split.
report_rows = []

for client_name in CLIENT_FILES:
    for split_name, (X, y) in raw_splits[client_name].items():
        X_prepared = normalize(X)

        np.savez_compressed(
            RUN_DIR / f"{client_name}_{split_name}.npz",
            X=X_prepared,
            y=y.astype(np.int64),
        )

        for class_id, records in pd.Series(y).value_counts().items():
            report_rows.append(
                {
                    "client": client_name,
                    "split": split_name,
                    "class_id": int(class_id),
                    "semantic_class": GLOBAL_CLASSES[int(class_id)],
                    "records": int(records),
                }
            )

        del X_prepared

    gc.collect()

report = pd.DataFrame(report_rows).sort_values(
    ["client", "split", "class_id"]
)
report.to_csv(RUN_DIR / "split_class_distribution.csv", index=False)

metadata = {
    "seed": SEED,
    "target": "category",
    "global_classes": GLOBAL_CLASSES,
    "class_to_id": CLASS_TO_ID,
    "category_to_semantic": CATEGORY_TO_SEMANTIC,
    "excluded_columns": sorted(FORBIDDEN_COLUMNS),
    "initial_feature_count": len(FEATURE_COLUMNS),
    "prepared_feature_count": len(SELECTED_FEATURES),
    "prepared_features": SELECTED_FEATURES,
    "max_rows_per_client": MAX_ROWS_PER_CLIENT,
    "sampling": "uniform-within-client; natural class imbalance retained",
    "normalization": "global mean/std from federated client training summaries only",
    "test_size_per_client": TEST_SIZE,
    "validation_size_per_client": VALIDATION_SIZE,
    "z_score_clip": Z_SCORE_CLIP,
}

with open(RUN_DIR / "metadata.json", "w") as file:
    json.dump(metadata, file, indent=2)

print(f"\nPrepared files saved in: {RUN_DIR}")
display(report.pivot_table(
    index=["client", "split"],
    columns="semantic_class",
    values="records",
    fill_value=0,
))
```

```python
# Cell 3 — Federated feature-selection warm-up
import json
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
from sklearn.feature_selection import mutual_info_classif
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split

warnings.filterwarnings("ignore", category=UserWarning)
warnings.filterwarnings("ignore", category=FutureWarning)

PREPARED_DIR = Path("/content/flmu_prepared")
FS_DIR = Path("/content/flmu_feature_selection")
FS_DIR.mkdir(exist_ok=True)

FEATURE_BUDGET = 20
FS_MAX_SAMPLES_PER_CLIENT = 30_000
SEED = 42

with open(PREPARED_DIR / "metadata.json") as file:
    metadata = json.load(file)

FEATURE_NAMES = metadata["prepared_features"]
CLIENTS = ["C1", "C2", "C3"]

assert len(FEATURE_NAMES) == 70, "Expected the 70 prepared features."


def representative_sample(X, y, maximum_rows):
    """Keeps the local class distribution while limiting feature-selection cost."""
    if len(y) <= maximum_rows:
        return X, y

    indices = np.arange(len(y))
    selected, _ = train_test_split(
        indices,
        train_size=maximum_rows,
        random_state=SEED,
        stratify=y,
    )
    return X[selected], y[selected]


def descending_rank(values):
    """Rank 1 means highest importance."""
    return pd.Series(values).rank(
        method="average",
        ascending=False
    ).to_numpy()


client_rank_tables = []
combined_rank_vectors = []

for client in CLIENTS:
    data = np.load(PREPARED_DIR / f"{client}_train.npz")
    X, y = data["X"], data["y"]

    X_fs, y_fs = representative_sample(X, y, FS_MAX_SAMPLES_PER_CLIENT)

    # Filter method: mutual information between feature and semantic category.
    mi_scores = mutual_info_classif(
        X_fs,
        y_fs,
        random_state=SEED,
        n_neighbors=3,
    )

    # Embedded method: sparse multinomial logistic regression.
    # balanced weights prevent dominant local classes from deciding all ranks.
    sparse_model = LogisticRegression(
        penalty="l1",
        solver="saga",
        multi_class="multinomial",
        class_weight="balanced",
        C=0.1,
        max_iter=300,
        random_state=SEED,
        n_jobs=-1,
    )
    sparse_model.fit(X_fs, y_fs)

    l1_scores = np.mean(np.abs(sparse_model.coef_), axis=0)

    mi_rank = descending_rank(mi_scores)
    l1_rank = descending_rank(l1_scores)

    # Equal hybrid weighting: lower combined rank is better.
    hybrid_rank = 0.5 * mi_rank + 0.5 * l1_rank
    combined_rank_vectors.append(hybrid_rank)

    client_table = pd.DataFrame({
        "client": client,
        "feature": FEATURE_NAMES,
        "mutual_information": mi_scores,
        "l1_importance": l1_scores,
        "mi_rank": mi_rank,
        "l1_rank": l1_rank,
        "hybrid_rank": hybrid_rank,
    }).sort_values("hybrid_rank")

    client_rank_tables.append(client_table)
    client_table.to_csv(FS_DIR / f"{client}_feature_ranks.csv", index=False)

    print(f"\n{client} top 10 local features:")
    display(client_table.head(10))

# Server-side rank aggregation: only rank values are combined.
rank_matrix = np.vstack(combined_rank_vectors)
server_rank = rank_matrix.mean(axis=0)

server_table = pd.DataFrame({
    "feature": FEATURE_NAMES,
    "server_mean_rank": server_rank,
    "C1_rank": rank_matrix[0],
    "C2_rank": rank_matrix[1],
    "C3_rank": rank_matrix[2],
}).sort_values("server_mean_rank").reset_index(drop=True)

selected_features = server_table.head(FEATURE_BUDGET)["feature"].tolist()
selected_indices = np.array(
    [FEATURE_NAMES.index(feature) for feature in selected_features],
    dtype=np.int64,
)

server_table.to_csv(FS_DIR / "server_aggregated_feature_ranks.csv", index=False)
np.save(FS_DIR / "selected_feature_indices.npy", selected_indices)

with open(FS_DIR / "selected_features.json", "w") as file:
    json.dump(
        {
            "feature_budget": FEATURE_BUDGET,
            "selected_features": selected_features,
            "selection_method": (
                "Client-side mutual-information and L1 ranks; "
                "server-side mean-rank aggregation"
            ),
            "feature_selection_samples_per_client": FS_MAX_SAMPLES_PER_CLIENT,
        },
        file,
        indent=2,
    )

print("\nShared top-20 feature schema:")
for number, feature in enumerate(selected_features, start=1):
    print(f"{number:02d}. {feature}")

display(server_table.head(FEATURE_BUDGET))
print(f"\nSaved feature-selection outputs to: {FS_DIR}")
```

```python
# Cell 4 — Train and evaluate Centralized, Local-only, FedAvg, and FedProx
# Test data are used ONLY for final evaluation. Model selection uses validation data.

import os
import json
import time
import copy
import random
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, TensorDataset

from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support,
    matthews_corrcoef,
    confusion_matrix,
)

# ----------------------- Configuration -----------------------

SEED = 42
BATCH_SIZE = 1024
LEARNING_RATE = 1e-3
WEIGHT_DECAY = 1e-4

FEDERATED_ROUNDS = 8
LOCAL_EPOCHS_PER_ROUND = 1
BASELINE_EPOCHS = 8

FOCAL_GAMMA = 2.0
FEDPROX_MU = 1e-3
MAX_CLASS_WEIGHT = 10.0

PREPARED_DIR = Path("/content/flmu_prepared")
FS_DIR = Path("/content/flmu_feature_selection")
RESULTS_DIR = Path("/content/flmu_results")
RESULTS_DIR.mkdir(exist_ok=True)

CLIENTS = ["C1", "C2", "C3"]

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
PIN_MEMORY = DEVICE.type == "cuda"

print("Using device:", DEVICE)

# ----------------------- Load prepared data -----------------------

with open(PREPARED_DIR / "metadata.json") as file:
    metadata = json.load(file)

with open(FS_DIR / "selected_features.json") as file:
    fs_metadata = json.load(file)

all_feature_names = metadata["prepared_features"]
selected_feature_names = fs_metadata["selected_features"]
selected_indices = np.load(FS_DIR / "selected_feature_indices.npy")

global_classes = metadata["global_classes"]
NUM_CLASSES = len(global_classes)
NORMAL_CLASS_ID = global_classes.index("Normal")

print("Model input features:", len(selected_feature_names))
print("Classes:", global_classes)

DATA = {}

for client in CLIENTS:
    DATA[client] = {}

    for split in ["train", "val", "test"]:
        data = np.load(PREPARED_DIR / f"{client}_{split}.npz")

        # Apply the globally agreed top-20 feature schema.
        X = data["X"][:, selected_indices].astype(np.float32)
        y = data["y"].astype(np.int64)

        DATA[client][split] = (X, y)

# ----------------------- Model and loss -----------------------

class LightweightMLP(nn.Module):
    def __init__(self, input_size, output_size):
        super().__init__()
        self.network = nn.Sequential(
            nn.Linear(input_size, 64),
            nn.ReLU(),
            nn.Dropout(0.20),
            nn.Linear(64, 32),
            nn.ReLU(),
            nn.Dropout(0.10),
            nn.Linear(32, output_size),
        )

    def forward(self, x):
        return self.network(x)


def model_size_bytes(model):
    return sum(
        parameter.numel() * parameter.element_size()
        for parameter in model.state_dict().values()
    )


def make_loader(X, y, shuffle):
    dataset = TensorDataset(
        torch.from_numpy(X),
        torch.from_numpy(y),
    )
    return DataLoader(
        dataset,
        batch_size=BATCH_SIZE,
        shuffle=shuffle,
        num_workers=0,
        pin_memory=PIN_MEMORY,
    )


def client_class_weights(y):
    """Inverse-frequency local weights, capped to prevent unstable rare-class loss."""
    counts = np.bincount(y, minlength=NUM_CLASSES).astype(np.float32)
    weights = np.zeros(NUM_CLASSES, dtype=np.float32)

    present = counts > 0
    if present.sum() == 0:
        raise ValueError("No labels found while computing class weights.")

    weights[present] = counts[present].sum() / (
        present.sum() * counts[present]
    )
    weights = np.clip(weights, 0.0, MAX_CLASS_WEIGHT)

    return torch.tensor(weights, dtype=torch.float32, device=DEVICE)


def focal_loss(logits, targets, weights, gamma=FOCAL_GAMMA):
    ce = F.cross_entropy(
        logits,
        targets,
        weight=weights,
        reduction="none",
    )
    probabilities = F.softmax(logits, dim=1)
    probability_of_true_class = probabilities.gather(
        1, targets.unsqueeze(1)
    ).squeeze(1)

    focal_term = (1.0 - probability_of_true_class).pow(gamma)
    return (focal_term * ce).mean()


def train_model(
    model,
    X,
    y,
    epochs,
    class_weights,
    reference_parameters=None,
    proximal_mu=0.0,
):
    """Trains one centralized or local model."""
    model = model.to(DEVICE)
    model.train()

    loader = make_loader(X, y, shuffle=True)

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=LEARNING_RATE,
        weight_decay=WEIGHT_DECAY,
    )

    epoch_losses = []

    for _ in range(epochs):
        loss_sum = 0.0
        records_seen = 0

        for X_batch, y_batch in loader:
            X_batch = X_batch.to(DEVICE, non_blocking=True)
            y_batch = y_batch.to(DEVICE, non_blocking=True)

            optimizer.zero_grad()

            logits = model(X_batch)
            loss = focal_loss(logits, y_batch, class_weights)

            # FedProx proximal regularization.
            if reference_parameters is not None and proximal_mu > 0:
                proximal_penalty = torch.zeros(1, device=DEVICE)

                for name, parameter in model.named_parameters():
                    proximal_penalty += torch.sum(
                        (parameter - reference_parameters[name]) ** 2
                    )

                loss = loss + (proximal_mu / 2.0) * proximal_penalty

            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=5.0)
            optimizer.step()

            loss_sum += loss.item() * len(y_batch)
            records_seen += len(y_batch)

        epoch_losses.append(loss_sum / max(records_seen, 1))

    return model.to("cpu"), float(np.mean(epoch_losses))


def evaluate_model(model, X, y):
    """Returns predictions and latency without retaining raw features."""
    model = model.to(DEVICE)
    model.eval()

    loader = make_loader(X, y, shuffle=False)
    predictions = []

    if DEVICE.type == "cuda":
        torch.cuda.synchronize()

    start_time = time.perf_counter()

    with torch.no_grad():
        for X_batch, _ in loader:
            X_batch = X_batch.to(DEVICE, non_blocking=True)
            predicted = torch.argmax(model(X_batch), dim=1)
            predictions.append(predicted.cpu().numpy())

    if DEVICE.type == "cuda":
        torch.cuda.synchronize()

    elapsed_seconds = time.perf_counter() - start_time
    predictions = np.concatenate(predictions)

    model.to("cpu")

    return y, predictions, elapsed_seconds


def calculate_metrics(y_true, y_pred):
    labels = list(range(NUM_CLASSES))

    precision, recall, f1, support = precision_recall_fscore_support(
        y_true,
        y_pred,
        labels=labels,
        zero_division=0,
    )

    macro_precision, macro_recall, macro_f1, _ = (
        precision_recall_fscore_support(
            y_true,
            y_pred,
            average="macro",
            zero_division=0,
        )
    )

    weighted_f1 = precision_recall_fscore_support(
        y_true,
        y_pred,
        average="weighted",
        zero_division=0,
    )[2]

    matrix = confusion_matrix(y_true, y_pred, labels=labels)

    # Benign-to-attack false-positive rate:
    # fraction of actual Normal records incorrectly predicted as any attack.
    normal_total = matrix[NORMAL_CLASS_ID].sum()
    benign_to_attack_fpr = (
        matrix[NORMAL_CLASS_ID].sum() - matrix[NORMAL_CLASS_ID, NORMAL_CLASS_ID]
    ) / normal_total if normal_total > 0 else np.nan

    summary = {
        "accuracy": accuracy_score(y_true, y_pred),
        "macro_precision": macro_precision,
        "macro_recall": macro_recall,
        "macro_f1": macro_f1,
        "weighted_f1": weighted_f1,
        "mcc": matthews_corrcoef(y_true, y_pred),
        "benign_to_attack_fpr": benign_to_attack_fpr,
    }

    per_class = pd.DataFrame({
        "class_id": labels,
        "class_name": global_classes,
        "precision": precision,
        "recall": recall,
        "f1": f1,
        "support": support,
    })

    return summary, per_class, matrix


def clone_state(model):
    return {
        key: value.detach().cpu().clone()
        for key, value in model.state_dict().items()
    }


def aggregate_models(local_models, local_sample_counts):
    """Sample-size-weighted FedAvg aggregation."""
    total_samples = float(sum(local_sample_counts))
    global_model = LightweightMLP(
        len(selected_feature_names),
        NUM_CLASSES,
    )

    aggregated_state = {}

    for key in global_model.state_dict().keys():
        weighted_value = sum(
            model.state_dict()[key].float() * (count / total_samples)
            for model, count in zip(local_models, local_sample_counts)
        )
        aggregated_state[key] = weighted_value.to(
            global_model.state_dict()[key].dtype
        )

    global_model.load_state_dict(aggregated_state)
    return global_model


def combined_validation(model):
    all_true, all_predicted = [], []

    for client in CLIENTS:
        X_val, y_val = DATA[client]["val"]
        y_true, y_pred, _ = evaluate_model(model, X_val, y_val)
        all_true.append(y_true)
        all_predicted.append(y_pred)

    summary, _, _ = calculate_metrics(
        np.concatenate(all_true),
        np.concatenate(all_predicted),
    )
    return summary


def run_federated(method_name, proximal_mu):
    global_model = LightweightMLP(
        len(selected_feature_names),
        NUM_CLASSES,
    )

    best_validation_f1 = -1.0
    best_state = clone_state(global_model)
    history = []

    start_time = time.perf_counter()

    for current_round in range(1, FEDERATED_ROUNDS + 1):
        local_models = []
        local_counts = []

        for client in CLIENTS:
            X_train, y_train = DATA[client]["train"]

            local_model = copy.deepcopy(global_model)
            local_weights = client_class_weights(y_train)

            reference = None
            if proximal_mu > 0:
                reference = {
                    name: parameter.detach().clone().to(DEVICE)
                    for name, parameter in local_model.named_parameters()
                }

            local_model, local_loss = train_model(
                local_model,
                X_train,
                y_train,
                epochs=LOCAL_EPOCHS_PER_ROUND,
                class_weights=local_weights,
                reference_parameters=reference,
                proximal_mu=proximal_mu,
            )

            local_models.append(local_model)
            local_counts.append(len(y_train))

            history.append({
                "method": method_name,
                "round": current_round,
                "client": client,
                "metric": "local_train_loss",
                "value": local_loss,
            })

        global_model = aggregate_models(local_models, local_counts)

        validation = combined_validation(global_model)

        history.append({
            "method": method_name,
            "round": current_round,
            "client": "AllClients",
            "metric": "validation_macro_f1",
            "value": validation["macro_f1"],
        })

        print(
            f"{method_name} | round {current_round:02d} | "
            f"validation macro-F1: {validation['macro_f1']:.4f}"
        )

        if validation["macro_f1"] > best_validation_f1:
            best_validation_f1 = validation["macro_f1"]
            best_state = clone_state(global_model)

    global_model.load_state_dict(best_state)
    elapsed = time.perf_counter() - start_time

    communication_bytes = (
        2 * len(CLIENTS) * FEDERATED_ROUNDS * model_size_bytes(global_model)
    )

    return global_model, pd.DataFrame(history), elapsed, communication_bytes


def evaluate_shared_model(method_name, model, elapsed, communication_bytes):
    summary_rows = []
    per_class_rows = []
    y_all, pred_all = [], []

    for client in CLIENTS:
        X_test, y_test = DATA[client]["test"]
        y_true, y_pred, inference_seconds = evaluate_model(model, X_test, y_test)

        metrics, per_class, _ = calculate_metrics(y_true, y_pred)

        summary_rows.append({
            "method": method_name,
            "evaluation_client": client,
            **metrics,
            "test_records": len(y_true),
            "inference_ms_per_record": 1000 * inference_seconds / len(y_true),
            "training_seconds": elapsed,
            "communication_rounds": FEDERATED_ROUNDS,
            "communication_mb": communication_bytes / (1024 ** 2),
            "parameters": sum(p.numel() for p in model.parameters()),
            "checkpoint_mb": model_size_bytes(model) / (1024 ** 2),
        })

        per_class["method"] = method_name
        per_class["evaluation_client"] = client
        per_class_rows.append(per_class)

        y_all.append(y_true)
        pred_all.append(y_pred)

    y_all = np.concatenate(y_all)
    pred_all = np.concatenate(pred_all)

    metrics, per_class, matrix = calculate_metrics(y_all, pred_all)

    summary_rows.append({
        "method": method_name,
        "evaluation_client": "AllClients",
        **metrics,
        "test_records": len(y_all),
        "inference_ms_per_record": np.nan,
        "training_seconds": elapsed,
        "communication_rounds": FEDERATED_ROUNDS,
        "communication_mb": communication_bytes / (1024 ** 2),
        "parameters": sum(p.numel() for p in model.parameters()),
        "checkpoint_mb": model_size_bytes(model) / (1024 ** 2),
    })

    per_class["method"] = method_name
    per_class["evaluation_client"] = "AllClients"
    per_class_rows.append(per_class)

    return summary_rows, per_class_rows, matrix


# ----------------------- Centralized baseline -----------------------

print("\n========== CENTRALIZED BASELINE ==========")

X_central = np.concatenate([DATA[client]["train"][0] for client in CLIENTS])
y_central = np.concatenate([DATA[client]["train"][1] for client in CLIENTS])

central_model = LightweightMLP(len(selected_feature_names), NUM_CLASSES)

start = time.perf_counter()
central_model, central_loss = train_model(
    central_model,
    X_central,
    y_central,
    epochs=BASELINE_EPOCHS,
    class_weights=client_class_weights(y_central),
)
central_time = time.perf_counter() - start

torch.save(
    central_model.state_dict(),
    RESULTS_DIR / "centralized_mlp.pt",
)

# ----------------------- Local-only baseline -----------------------

print("\n========== LOCAL-ONLY BASELINE ==========")

local_models = {}
local_training_time = 0.0

for client in CLIENTS:
    X_train, y_train = DATA[client]["train"]

    model = LightweightMLP(len(selected_feature_names), NUM_CLASSES)

    start = time.perf_counter()
    model, local_loss = train_model(
        model,
        X_train,
        y_train,
        epochs=BASELINE_EPOCHS,
        class_weights=client_class_weights(y_train),
    )
    local_training_time += time.perf_counter() - start

    local_models[client] = model
    torch.save(
        model.state_dict(),
        RESULTS_DIR / f"local_only_{client}.pt",
    )

# ----------------------- Federated experiments -----------------------

print("\n========== FEDAVG ==========")
fedavg_model, fedavg_history, fedavg_time, fedavg_bytes = run_federated(
    "FedAvg",
    proximal_mu=0.0,
)
torch.save(fedavg_model.state_dict(), RESULTS_DIR / "fedavg_mlp.pt")

print("\n========== PROPOSED FEDPROX ==========")
fedprox_model, fedprox_history, fedprox_time, fedprox_bytes = run_federated(
    "FedProx_Focal_Top20",
    proximal_mu=FEDPROX_MU,
)
torch.save(fedprox_model.state_dict(), RESULTS_DIR / "fedprox_focal_top20_mlp.pt")

# ----------------------- Final testing -----------------------

summary_rows = []
per_class_rows = []
matrices = {}

rows, classes, matrix = evaluate_shared_model(
    "Centralized",
    central_model,
    central_time,
    communication_bytes=0,
)
summary_rows.extend(rows)
per_class_rows.extend(classes)
matrices["Centralized"] = matrix

# Local-only model: evaluate each local model on only its own local test set.
local_y_all, local_pred_all = [], []

for client in CLIENTS:
    X_test, y_test = DATA[client]["test"]
    y_true, y_pred, inference_seconds = evaluate_model(
        local_models[client],
        X_test,
        y_test,
    )

    metrics, per_class, _ = calculate_metrics(y_true, y_pred)

    summary_rows.append({
        "method": "LocalOnly",
        "evaluation_client": client,
        **metrics,
        "test_records": len(y_true),
        "inference_ms_per_record": 1000 * inference_seconds / len(y_true),
        "training_seconds": local_training_time,
        "communication_rounds": 0,
        "communication_mb": 0.0,
        "parameters": sum(p.numel() for p in local_models[client].parameters()),
        "checkpoint_mb": model_size_bytes(local_models[client]) / (1024 ** 2),
    })

    per_class["method"] = "LocalOnly"
    per_class["evaluation_client"] = client
    per_class_rows.append(per_class)

    local_y_all.append(y_true)
    local_pred_all.append(y_pred)

local_metrics, local_per_class, local_matrix = calculate_metrics(
    np.concatenate(local_y_all),
    np.concatenate(local_pred_all),
)

summary_rows.append({
    "method": "LocalOnly",
    "evaluation_client": "AllClients",
    **local_metrics,
    "test_records": sum(len(values) for values in local_y_all),
    "inference_ms_per_record": np.nan,
    "training_seconds": local_training_time,
    "communication_rounds": 0,
    "communication_mb": 0.0,
    "parameters": sum(p.numel() for p in local_models["C1"].parameters()),
    "checkpoint_mb": model_size_bytes(local_models["C1"]) / (1024 ** 2),
})

local_per_class["method"] = "LocalOnly"
local_per_class["evaluation_client"] = "AllClients"
per_class_rows.append(local_per_class)
matrices["LocalOnly"] = local_matrix

rows, classes, matrix = evaluate_shared_model(
    "FedAvg",
    fedavg_model,
    fedavg_time,
    fedavg_bytes,
)
summary_rows.extend(rows)
per_class_rows.extend(classes)
matrices["FedAvg"] = matrix

rows, classes, matrix = evaluate_shared_model(
    "FedProx_Focal_Top20",
    fedprox_model,
    fedprox_time,
    fedprox_bytes,
)
summary_rows.extend(rows)
per_class_rows.extend(classes)
matrices["FedProx_Focal_Top20"] = matrix

# ----------------------- Save factual outputs -----------------------

summary = pd.DataFrame(summary_rows)
per_class = pd.concat(per_class_rows, ignore_index=True)
history = pd.concat([fedavg_history, fedprox_history], ignore_index=True)

summary.to_csv(RESULTS_DIR / "main_results.csv", index=False)
per_class.to_csv(RESULTS_DIR / "per_class_results.csv", index=False)
history.to_csv(RESULTS_DIR / "training_history.csv", index=False)

run_config = {
    "seed": SEED,
    "device": str(DEVICE),
    "selected_features": selected_feature_names,
    "model": "MLP(20 -> 64 -> 32 -> 7)",
    "batch_size": BATCH_SIZE,
    "learning_rate": LEARNING_RATE,
    "focal_gamma": FOCAL_GAMMA,
    "maximum_class_weight": MAX_CLASS_WEIGHT,
    "fedprox_mu": FEDPROX_MU,
    "federated_rounds": FEDERATED_ROUNDS,
    "local_epochs_per_round": LOCAL_EPOCHS_PER_ROUND,
    "baseline_epochs": BASELINE_EPOCHS,
    "methods": [
        "Centralized",
        "LocalOnly",
        "FedAvg",
        "FedProx_Focal_Top20",
    ],
    "note": (
        "Results use a reproducible 150,000-record maximum natural-distribution "
        "sample per client. They must be reported as sampled-client experiments."
    ),
}

with open(RESULTS_DIR / "run_config.json", "w") as file:
    json.dump(run_config, file, indent=2)

# ----------------------- Figures -----------------------

sns.set_theme(style="whitegrid")

for method, matrix in matrices.items():
    plt.figure(figsize=(9, 7))
    sns.heatmap(
        matrix,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=global_classes,
        yticklabels=global_classes,
    )
    plt.xlabel("Predicted class")
    plt.ylabel("True class")
    plt.title(f"Confusion Matrix — {method}")
    plt.tight_layout()
    plt.savefig(
        RESULTS_DIR / f"confusion_matrix_{method}.png",
        dpi=300,
        bbox_inches="tight",
    )
    plt.show()
    plt.close()

all_client_results = summary[
    summary["evaluation_client"] == "AllClients"
].copy()

plot_columns = ["accuracy", "macro_f1", "mcc", "benign_to_attack_fpr"]

plt.figure(figsize=(11, 5))
melted = all_client_results.melt(
    id_vars="method",
    value_vars=plot_columns,
    var_name="metric",
    value_name="value",
)
sns.barplot(data=melted, x="metric", y="value", hue="method")
plt.ylim(0, 1.05)
plt.title("Final Test Metrics Across All Client Test Partitions")
plt.tight_layout()
plt.savefig(RESULTS_DIR / "main_metric_comparison.png", dpi=300)
plt.show()
plt.close()

validation_history = history[
    history["metric"] == "validation_macro_f1"
].copy()

plt.figure(figsize=(9, 5))
sns.lineplot(
    data=validation_history,
    x="round",
    y="value",
    hue="method",
    marker="o",
)
plt.ylim(0, 1.05)
plt.ylabel("Validation Macro-F1")
plt.title("Federated Validation Performance")
plt.tight_layout()
plt.savefig(RESULTS_DIR / "federated_validation_macro_f1.png", dpi=300)
plt.show()
plt.close()

print("\n========== FINAL ALL-CLIENT RESULTS ==========")
display(
    all_client_results[
        [
            "method",
            "accuracy",
            "macro_precision",
            "macro_recall",
            "macro_f1",
            "weighted_f1",
            "mcc",
            "benign_to_attack_fpr",
            "training_seconds",
            "communication_rounds",
            "communication_mb",
            "parameters",
            "checkpoint_mb",
        ]
    ].sort_values("macro_f1", ascending=False)
)

print(f"\nAll factual outputs are saved in: {RESULTS_DIR}")
```

```python
# Cell 5 — Client-Balanced Personalized FedProx (CB-PFedProx)
# Requires the variables/functions from Cell 4 to remain available.

CB_ROUNDS = 15
CB_FEDPROX_MU = 0.01
PERSONALIZATION_EPOCHS = 3
IMPROVED_DIR = RESULTS_DIR / "cb_pfedprox"
IMPROVED_DIR.mkdir(exist_ok=True)

def aggregate_equal_clients(local_models):
    """Each client contributes equally, preventing C2 from dominating aggregation."""
    global_model = LightweightMLP(
        len(selected_feature_names),
        NUM_CLASSES,
    )

    aggregated_state = {}

    for key in global_model.state_dict().keys():
        values = [
            local_model.state_dict()[key].float()
            for local_model in local_models
        ]
        aggregated_state[key] = torch.stack(values, dim=0).mean(dim=0).to(
            global_model.state_dict()[key].dtype
        )

    global_model.load_state_dict(aggregated_state)
    return global_model


def run_client_balanced_fedprox():
    global_model = LightweightMLP(
        len(selected_feature_names),
        NUM_CLASSES,
    )

    best_validation_f1 = -1.0
    best_state = clone_state(global_model)
    history = []
    start_time = time.perf_counter()

    for round_number in range(1, CB_ROUNDS + 1):
        local_models = []

        for client in CLIENTS:
            X_train, y_train = DATA[client]["train"]

            local_model = copy.deepcopy(global_model)
            local_weights = client_class_weights(y_train)

            reference_parameters = {
                name: parameter.detach().clone().to(DEVICE)
                for name, parameter in local_model.named_parameters()
            }

            local_model, local_loss = train_model(
                local_model,
                X_train,
                y_train,
                epochs=1,
                class_weights=local_weights,
                reference_parameters=reference_parameters,
                proximal_mu=CB_FEDPROX_MU,
            )

            local_models.append(local_model)

            history.append({
                "phase": "federated_training",
                "round": round_number,
                "client": client,
                "metric": "local_train_loss",
                "value": local_loss,
            })

        global_model = aggregate_equal_clients(local_models)

        validation = combined_validation(global_model)

        history.append({
            "phase": "federated_training",
            "round": round_number,
            "client": "AllClients",
            "metric": "validation_macro_f1",
            "value": validation["macro_f1"],
        })

        print(
            f"CB-PFedProx | Round {round_number:02d} | "
            f"Validation macro-F1: {validation['macro_f1']:.4f}"
        )

        if validation["macro_f1"] > best_validation_f1:
            best_validation_f1 = validation["macro_f1"]
            best_state = clone_state(global_model)

    global_model.load_state_dict(best_state)

    return global_model, pd.DataFrame(history), time.perf_counter() - start_time


def personalize_for_client(global_model, client):
    """
    Local adaptation uses only local training and validation data.
    Test data are never used for adaptation or model selection.
    """
    X_train, y_train = DATA[client]["train"]
    X_val, y_val = DATA[client]["val"]

    local_model = copy.deepcopy(global_model)
    weights = client_class_weights(y_train)

    best_state = clone_state(local_model)
    best_f1 = -1.0
    personal_history = []

    for epoch in range(1, PERSONALIZATION_EPOCHS + 1):
        local_model, loss = train_model(
            local_model,
            X_train,
            y_train,
            epochs=1,
            class_weights=weights,
            reference_parameters=None,
            proximal_mu=0.0,
        )

        y_true, y_pred, _ = evaluate_model(local_model, X_val, y_val)
        validation, _, _ = calculate_metrics(y_true, y_pred)

        personal_history.append({
            "phase": "personalization",
            "round": epoch,
            "client": client,
            "metric": "validation_macro_f1",
            "value": validation["macro_f1"],
        })

        print(
            f"{client} personalization | epoch {epoch} | "
            f"validation macro-F1: {validation['macro_f1']:.4f}"
        )

        if validation["macro_f1"] > best_f1:
            best_f1 = validation["macro_f1"]
            best_state = clone_state(local_model)

    local_model.load_state_dict(best_state)
    return local_model, pd.DataFrame(personal_history)


# 1. Global client-balanced FedProx training.
cb_global_model, cb_history, cb_global_seconds = run_client_balanced_fedprox()

# 2. Personalize the best global model locally.
personalized_models = {}
personalization_histories = []
personalization_seconds = 0.0

for client in CLIENTS:
    start_time = time.perf_counter()

    personalized_model, client_history = personalize_for_client(
        cb_global_model,
        client,
    )

    personalization_seconds += time.perf_counter() - start_time
    personalized_models[client] = personalized_model
    personalization_histories.append(client_history)

    torch.save(
        personalized_model.state_dict(),
        IMPROVED_DIR / f"cb_pfedprox_personalized_{client}.pt",
    )

torch.save(
    cb_global_model.state_dict(),
    IMPROVED_DIR / "cb_pfedprox_global_model.pt",
)

# 3. Final local-client test evaluation.
summary_rows = []
per_class_rows = []
all_true, all_pred = [], []

total_training_seconds = cb_global_seconds + personalization_seconds
model_bytes = model_size_bytes(cb_global_model)
communication_bytes = 2 * len(CLIENTS) * CB_ROUNDS * model_bytes

for client in CLIENTS:
    X_test, y_test = DATA[client]["test"]

    y_true, y_pred, inference_seconds = evaluate_model(
        personalized_models[client],
        X_test,
        y_test,
    )

    metrics, per_class, _ = calculate_metrics(y_true, y_pred)

    summary_rows.append({
        "method": "CB_PFedProx_Personalized",
        "evaluation_client": client,
        **metrics,
        "test_records": len(y_true),
        "inference_ms_per_record": 1000 * inference_seconds / len(y_true),
        "training_seconds": total_training_seconds,
        "communication_rounds": CB_ROUNDS,
        "communication_mb": communication_bytes / (1024 ** 2),
        "parameters": sum(
            parameter.numel()
            for parameter in personalized_models[client].parameters()
        ),
        "checkpoint_mb": model_size_bytes(
            personalized_models[client]
        ) / (1024 ** 2),
    })

    per_class["method"] = "CB_PFedProx_Personalized"
    per_class["evaluation_client"] = client
    per_class_rows.append(per_class)

    all_true.append(y_true)
    all_pred.append(y_pred)

# All-client aggregate metrics.
all_true = np.concatenate(all_true)
all_pred = np.concatenate(all_pred)

all_metrics, all_per_class, all_matrix = calculate_metrics(all_true, all_pred)

summary_rows.append({
    "method": "CB_PFedProx_Personalized",
    "evaluation_client": "AllClients",
    **all_metrics,
    "test_records": len(all_true),
    "inference_ms_per_record": np.nan,
    "training_seconds": total_training_seconds,
    "communication_rounds": CB_ROUNDS,
    "communication_mb": communication_bytes / (1024 ** 2),
    "parameters": sum(parameter.numel() for parameter in cb_global_model.parameters()),
    "checkpoint_mb": model_bytes / (1024 ** 2),
})

all_per_class["method"] = "CB_PFedProx_Personalized"
all_per_class["evaluation_client"] = "AllClients"
per_class_rows.append(all_per_class)

# Save results.
cb_summary = pd.DataFrame(summary_rows)
cb_per_class = pd.concat(per_class_rows, ignore_index=True)
cb_full_history = pd.concat(
    [cb_history] + personalization_histories,
    ignore_index=True,
)

cb_summary.to_csv(IMPROVED_DIR / "cb_pfedprox_results.csv", index=False)
cb_per_class.to_csv(IMPROVED_DIR / "cb_pfedprox_per_class_results.csv", index=False)
cb_full_history.to_csv(IMPROVED_DIR / "cb_pfedprox_history.csv", index=False)

# Confusion matrix.
plt.figure(figsize=(9, 7))
sns.heatmap(
    all_matrix,
    annot=True,
    fmt="d",
    cmap="Greens",
    xticklabels=global_classes,
    yticklabels=global_classes,
)
plt.xlabel("Predicted class")
plt.ylabel("True class")
plt.title("Confusion Matrix — CB-PFedProx Personalized")
plt.tight_layout()
plt.savefig(
    IMPROVED_DIR / "cb_pfedprox_confusion_matrix.png",
    dpi=300,
    bbox_inches="tight",
)
plt.show()
plt.close()

# Compare against existing all-client baselines.
baseline_results = pd.read_csv(RESULTS_DIR / "main_results.csv")
baseline_results = baseline_results[
    baseline_results["evaluation_client"] == "AllClients"
]

comparison = pd.concat(
    [baseline_results, cb_summary[cb_summary["evaluation_client"] == "AllClients"]],
    ignore_index=True,
)

comparison.to_csv(
    IMPROVED_DIR / "cb_pfedprox_comparison.csv",
    index=False,
)

print("\n========== COMPARISON ==========")
display(
    comparison[
        [
            "method",
            "accuracy",
            "macro_precision",
            "macro_recall",
            "macro_f1",
            "weighted_f1",
            "mcc",
            "benign_to_attack_fpr",
            "communication_rounds",
            "communication_mb",
        ]
    ].sort_values("macro_f1", ascending=False)
)

print(f"\nSaved CB-PFedProx outputs to: {IMPROVED_DIR}")
```

```python
# Cell 5A — Restore the training environment for CB-PFedProx
import json
import time
import copy
import random
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, TensorDataset

from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support,
    matthews_corrcoef,
    confusion_matrix,
)

SEED = 42
BATCH_SIZE = 1024
LEARNING_RATE = 1e-3
WEIGHT_DECAY = 1e-4
FOCAL_GAMMA = 2.0
MAX_CLASS_WEIGHT = 10.0

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
PIN_MEMORY = DEVICE.type == "cuda"

PREPARED_DIR = Path("/content/flmu_prepared")
FS_DIR = Path("/content/flmu_feature_selection")
RESULTS_DIR = Path("/content/flmu_results")
RESULTS_DIR.mkdir(exist_ok=True)

CLIENTS = ["C1", "C2", "C3"]

with open(PREPARED_DIR / "metadata.json") as file:
    metadata = json.load(file)

with open(FS_DIR / "selected_features.json") as file:
    fs_metadata = json.load(file)

selected_feature_names = fs_metadata["selected_features"]
selected_indices = np.load(FS_DIR / "selected_feature_indices.npy")
global_classes = metadata["global_classes"]
NUM_CLASSES = len(global_classes)
NORMAL_CLASS_ID = global_classes.index("Normal")

DATA = {}

for client in CLIENTS:
    DATA[client] = {}

    for split in ["train", "val", "test"]:
        saved = np.load(PREPARED_DIR / f"{client}_{split}.npz")

        DATA[client][split] = (
            saved["X"][:, selected_indices].astype(np.float32),
            saved["y"].astype(np.int64),
        )

print("Restored data and functions.")
print("Device:", DEVICE)
print("Input features:", len(selected_feature_names))
print("Classes:", global_classes)


class LightweightMLP(nn.Module):
    def __init__(self, input_size, output_size):
        super().__init__()
        self.network = nn.Sequential(
            nn.Linear(input_size, 64),
            nn.ReLU(),
            nn.Dropout(0.20),
            nn.Linear(64, 32),
            nn.ReLU(),
            nn.Dropout(0.10),
            nn.Linear(32, output_size),
        )

    def forward(self, x):
        return self.network(x)


def model_size_bytes(model):
    return sum(
        parameter.numel() * parameter.element_size()
        for parameter in model.state_dict().values()
    )


def make_loader(X, y, shuffle):
    dataset = TensorDataset(
        torch.from_numpy(X),
        torch.from_numpy(y),
    )

    return DataLoader(
        dataset,
        batch_size=BATCH_SIZE,
        shuffle=shuffle,
        num_workers=0,
        pin_memory=PIN_MEMORY,
    )


def client_class_weights(y):
    counts = np.bincount(y, minlength=NUM_CLASSES).astype(np.float32)
    weights = np.zeros(NUM_CLASSES, dtype=np.float32)
    present = counts > 0

    weights[present] = counts[present].sum() / (
        present.sum() * counts[present]
    )
    weights = np.clip(weights, 0.0, MAX_CLASS_WEIGHT)

    return torch.tensor(weights, dtype=torch.float32, device=DEVICE)


def focal_loss(logits, targets, weights, gamma=FOCAL_GAMMA):
    cross_entropy = F.cross_entropy(
        logits,
        targets,
        weight=weights,
        reduction="none",
    )

    probabilities = F.softmax(logits, dim=1)
    true_class_probability = probabilities.gather(
        1, targets.unsqueeze(1)
    ).squeeze(1)

    return ((1.0 - true_class_probability).pow(gamma) * cross_entropy).mean()


def train_model(
    model,
    X,
    y,
    epochs,
    class_weights,
    reference_parameters=None,
    proximal_mu=0.0,
):
    model = model.to(DEVICE)
    model.train()

    loader = make_loader(X, y, shuffle=True)

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=LEARNING_RATE,
        weight_decay=WEIGHT_DECAY,
    )

    all_epoch_losses = []

    for _ in range(epochs):
        total_loss = 0.0
        total_records = 0

        for X_batch, y_batch in loader:
            X_batch = X_batch.to(DEVICE, non_blocking=True)
            y_batch = y_batch.to(DEVICE, non_blocking=True)

            optimizer.zero_grad()

            logits = model(X_batch)
            loss = focal_loss(logits, y_batch, class_weights)

            if reference_parameters is not None and proximal_mu > 0:
                proximal_penalty = torch.zeros(1, device=DEVICE)

                for name, parameter in model.named_parameters():
                    proximal_penalty += torch.sum(
                        (parameter - reference_parameters[name]) ** 2
                    )

                loss = loss + (proximal_mu / 2.0) * proximal_penalty

            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), 5.0)
            optimizer.step()

            total_loss += loss.item() * len(y_batch)
            total_records += len(y_batch)

        all_epoch_losses.append(total_loss / max(total_records, 1))

    return model.to("cpu"), float(np.mean(all_epoch_losses))


def evaluate_model(model, X, y):
    model = model.to(DEVICE)
    model.eval()

    loader = make_loader(X, y, shuffle=False)
    predictions = []

    if DEVICE.type == "cuda":
        torch.cuda.synchronize()

    start_time = time.perf_counter()

    with torch.no_grad():
        for X_batch, _ in loader:
            X_batch = X_batch.to(DEVICE, non_blocking=True)
            predicted = torch.argmax(model(X_batch), dim=1)
            predictions.append(predicted.cpu().numpy())

    if DEVICE.type == "cuda":
        torch.cuda.synchronize()

    elapsed_seconds = time.perf_counter() - start_time
    model.to("cpu")

    return y, np.concatenate(predictions), elapsed_seconds


def calculate_metrics(y_true, y_pred):
    labels = list(range(NUM_CLASSES))

    precision, recall, f1, support = precision_recall_fscore_support(
        y_true, y_pred, labels=labels, zero_division=0
    )

    macro_precision, macro_recall, macro_f1, _ = (
        precision_recall_fscore_support(
            y_true, y_pred, average="macro", zero_division=0
        )
    )

    weighted_f1 = precision_recall_fscore_support(
        y_true, y_pred, average="weighted", zero_division=0
    )[2]

    matrix = confusion_matrix(y_true, y_pred, labels=labels)

    normal_total = matrix[NORMAL_CLASS_ID].sum()
    fpr = (
        matrix[NORMAL_CLASS_ID].sum() - matrix[NORMAL_CLASS_ID, NORMAL_CLASS_ID]
    ) / normal_total if normal_total > 0 else np.nan

    metrics = {
        "accuracy": accuracy_score(y_true, y_pred),
        "macro_precision": macro_precision,
        "macro_recall": macro_recall,
        "macro_f1": macro_f1,
        "weighted_f1": weighted_f1,
        "mcc": matthews_corrcoef(y_true, y_pred),
        "benign_to_attack_fpr": fpr,
    }

    per_class = pd.DataFrame({
        "class_id": labels,
        "class_name": global_classes,
        "precision": precision,
        "recall": recall,
        "f1": f1,
        "support": support,
    })

    return metrics, per_class, matrix


def clone_state(model):
    return {
        key: value.detach().cpu().clone()
        for key, value in model.state_dict().items()
    }


def combined_validation(model):
    all_true, all_predicted = [], []

    for client in CLIENTS:
        X_val, y_val = DATA[client]["val"]
        y_true, y_pred, _ = evaluate_model(model, X_val, y_val)

        all_true.append(y_true)
        all_predicted.append(y_pred)

    metrics, _, _ = calculate_metrics(
        np.concatenate(all_true),
        np.concatenate(all_predicted),
    )

    return metrics
```
